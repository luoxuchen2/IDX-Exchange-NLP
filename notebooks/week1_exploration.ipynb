```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("../data/processed/listing_sample.csv")

# Basic exploration
print("Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nRemark Length Summary:")
df["remark_length"] = df["remarks"].str.len()
print(df["remark_length"].describe())

print("\nTop 10 Cities:")
print(df["L_City"].value_counts().head(10))

print("\nPrice Summary:")
print(df["price"].describe())

print("\nBeds and Baths Summary:")
print(df[["beds", "baths"]].describe())


# Preview data
df.head()
```

```python

# Plot remark lengths
df["remark_length"].hist(bins=30)
plt.xlabel("Remark Length")
plt.ylabel("Number of Listings")
plt.title("Distribution of Listing Remark Length")
plt.show()
```
